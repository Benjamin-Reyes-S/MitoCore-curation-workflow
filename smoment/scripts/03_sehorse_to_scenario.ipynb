# Preparation of Seahorse data for MitoCore

This script parses the seahorse data from the input excel file (seahorse_data.xlsx) and write scenario files for sMOMENT. These files are used to determine reaction candidates for kcat calibration.

```python
import pandas
import json
```

```python
seahorse_df = pandas.read_excel('./../model_data_input/seahorse_data.xlsx')
```

```python
seahorse_df
```

```python
flux_data = {
    'platelet': {
        'target' : {
            'reaction': 'R_' + str(seahorse_df.loc[seahorse_df['parameter'] == 'ATP synthesis rate max', 'model_reaction'].values[0]),
            'value': seahorse_df.loc[seahorse_df['parameter'] == 'ATP synthesis rate max', 'value_mmol_gdw_h'].values[0]
        },
        'setup' : {}
        }
    }

with open('./../model_data_input/platelet_scenario.json', 'w') as fp:
                json.dump(flux_data, fp)
```
