```python
Mito_MA = cobra.io.read_sbml_model('MitoCore_Human1_MA.xml')
Mito_Last = cobra.io.read_sbml_model('/Users/benjaminreyes/Desktop/Masterarbeit/MitoCore_for_Disease_Modelling/Output_Models_MitoCore/Mitocore_aligned_to_Human1.xml')
```

check Uniprot

```python
uniprot_last={}

for gene in Mito_Last.genes:
    annotations= gene.annotation
    if 'uniprot' in annotations.keys():
        uniprot_id = annotations['uniprot']
        uniprot_last[gene.id]=uniprot_id
print("uniprot_last:", uniprot_last)
uniprot_MA={}
for gene in Mito_MA.genes:
    annotations= gene.annotation
    if 'uniprot' in annotations.keys():
        uniprot_id = annotations['uniprot']
        uniprot_MA[gene.id]=uniprot_id



for id, uniprot in uniprot_last.items():
    if id in uniprot_MA.keys():
        if uniprot_MA[id] != uniprot:
            print(f"Gene {id} has different uniprot ids: {uniprot_MA[id]} and {uniprot}")

```

check BiGG ids

```python
bigg_last={}
for met in Mito_Last.metabolites:
    annotations= met.annotation
    if 'bigg.metabolite' in annotations.keys():
        bigg_id = annotations['bigg.metabolite']
        bigg_last[met.id]=bigg_id
print("bigg_last:", bigg_last)

bigg_MA={}
for met in Mito_MA.metabolites:
    annotations= met.annotation
    if 'bigg.metabolite' in annotations.keys():
        bigg_id = annotations['bigg.metabolite']
        bigg_MA[met.id]=bigg_id
print("bigg_MA:", bigg_MA)


for id, bigg in bigg_last.items():
    if id in bigg_MA.keys():
        if bigg_MA[id] != bigg:
            print(f"Metabolite {id} has different bigg ids, bigg MA: {bigg_MA[id]} and bigg last: {bigg}")
        #else: print(f"Metabolite {id} has the same bigg id: {bigg_MA[id]} and {bigg}")
```

```python
gpr_last={}

for reaction in Mito_Last.reactions:
    gpr= reaction.gene_reaction_rule
    gpr_last[reaction.id]=gpr

gpr_MA={}
for reaction in Mito_MA.reactions:
    gpr= reaction.gene_reaction_rule
    gpr_MA[reaction.id]=gpr



for id, gpr in gpr_last.items():
    if id in gpr_MA.keys():
        if gpr_MA[id] != gpr:
            print(f"Reaction {id} has different GPRs, gpr MA: {gpr_MA[id]} and gpr last: {gpr}")
```

```python
ec_last={}

for reaction in Mito_Last.reactions:
    annotations= reaction.annotation
    ec= annotations.get('ec-code', None)
    ec_last[reaction.id]=ec
ec_MA={}
for reaction in Mito_MA.reactions:
    annotations= reaction.annotation
    ec= annotations.get('ec-code', None)
    ec_MA[reaction.id]=ec

for id, ec in ec_last.items():
    if id in ec_MA.keys():
        if ec_MA[id] != ec:
            print(f"Reaction {id} has different EC numbers, ec MA: {ec_MA[id]} and ec last: {ec}")
```
